```python
import math
import random
from IPython.display import display, clear_output
import ipywidgets as widgets
```

```python
# Create an empty Tic-Tac-Toe board
board = [" " for _ in range(9)]

# Winning combinations
WINNING_COMBINATIONS = [
    (0, 1, 2),
    (3, 4, 5),
    (6, 7, 8),
    (0, 3, 6),
    (1, 4, 7),
    (2, 5, 8),
    (0, 4, 8),
    (2, 4, 6)
]

# Game status
game_over = False
message = "Your turn! You are X."
```

```python
def check_winner(board):
    """
    Check whether X or O has won.
    Returns:
        'X' if player wins
        'O' if AI wins
        'Draw' if board is full
        None if game is still running
    """

    for a, b, c in WINNING_COMBINATIONS:
        if board[a] != " " and board[a] == board[b] == board[c]:
            return board[a]

    if " " not in board:
        return "Draw"

    return None
```

```python
def minimax(board, depth, is_maximizing):
    """
    Minimax algorithm for the AI.

    O = AI
    X = Human player

    The AI tries to maximize its score,
    while assuming that the human will make
    the best possible move.
    """

    result = check_winner(board)

    # Terminal states
    if result == "O":
        return 10 - depth

    if result == "X":
        return depth - 10

    if result == "Draw":
        return 0

    # AI's turn - maximize score
    if is_maximizing:
        best_score = -math.inf

        for i in range(9):
            if board[i] == " ":
                board[i] = "O"

                score = minimax(board, depth + 1, False)

                board[i] = " "
                best_score = max(best_score, score)

        return best_score

    # Human's turn - minimize score
    else:
        best_score = math.inf

        for i in range(9):
            if board[i] == " ":
                board[i] = "X"

                score = minimax(board, depth + 1, True)

                board[i] = " "
                best_score = min(best_score, score)

        return best_score
```

```python
def get_ai_move(board):
    """
    Find the best possible move for the AI using Minimax.
    """

    best_score = -math.inf
    best_move = None

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"

            score = minimax(board, 0, False)

            board[i] = " "

            if score > best_score:
                best_score = score
                best_move = i

    return best_move
```

```python
def display_board():
    clear_output(wait=True)

    print("================================")
    print("       AI TIC-TAC-TOE")
    print("================================")
    print()
    print("You = X     Computer AI = O")
    print()

    for i in range(0, 9, 3):
        print(f"  {board[i]}  |  {board[i+1]}  |  {board[i+2]}")

        if i < 6:
            print("-----+-----+-----")

    print()
    print(message)
    print()
```

```python
buttons = []

for i in range(9):
    button = widgets.Button(
        description=" ",
        layout=widgets.Layout(width="70px", height="70px")
    )

    buttons.append(button)

grid = widgets.GridBox(
    buttons,
    layout=widgets.Layout(
        grid_template_columns="repeat(3, 70px)",
        grid_template_rows="repeat(3, 70px)",
        grid_gap="5px"
    )
)
```

```python
def player_move(index):
    global board, game_over, message

    # Ignore move if game has ended
    if game_over:
        return

    # Ignore occupied position
    if board[index] != " ":
        message = "That position is already occupied!"
        display_board()
        display(grid)
        display(reset_button)
        return

    # Human move
    board[index] = "X"

    result = check_winner(board)

    if result == "X":
        message = "Congratulations! You won!"
        game_over = True
        update_buttons()
        display_board()
        display(grid)
        display(reset_button)
        return

    if result == "Draw":
        message = "It's a draw!"
        game_over = True
        update_buttons()
        display_board()
        display(grid)
        display(reset_button)
        return

    # AI move
    message = "AI is thinking..."
    update_buttons()
    display_board()
    display(grid)

    ai_move = get_ai_move(board)

    if ai_move is not None:
        board[ai_move] = "O"

    # Check result after AI move
    result = check_winner(board)

    if result == "O":
        message = "AI wins! Try again."
        game_over = True

    elif result == "Draw":
        message = "It's a draw!"
        game_over = True

    else:
        message = "Your turn! You are X."

    update_buttons()
    display_board()
    display(grid)
    display(reset_button)
```

```python
def update_buttons():
    for i in range(9):
        buttons[i].description = board[i]

        if board[i] != " ":
            buttons[i].disabled = True
        else:
            buttons[i].disabled = game_over


def reset_game(b=None):
    global board, game_over, message

    board = [" " for _ in range(9)]
    game_over = False
    message = "Your turn! You are X."

    for button in buttons:
        button.description = " "
        button.disabled = False

    display_game()


def display_game():
    clear_output(wait=True)

    print("================================")
    print("       AI TIC-TAC-TOE")
    print("================================")
    print()
    print("You = X     Computer AI = O")
    print()
    print(message)
    print()

    display(grid)
    display(reset_button)
```

```python
for i, button in enumerate(buttons):
    button.on_click(lambda b, i=i: player_move(i))

reset_button = widgets.Button(
    description="Restart Game",
    button_style="info",
    layout=widgets.Layout(width="150px", height="40px")
)

reset_button.on_click(reset_game)

display_game()
```
